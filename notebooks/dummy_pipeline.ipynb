# Dummy Medallion Pipeline

This notebook demonstrates a simple medallion pipeline: ingest raw data, transform to bronze, enrich to silver, and aggregate to gold.

```python
import os
import pandas as pd

base = os.path.join('..','data')
raw = os.path.join(base,'raw')
bronze = os.path.join(base,'bronze')
silver = os.path.join(base,'silver')
gold = os.path.join(base,'gold')

# create example raw csv
df = pd.DataFrame({'id':[1,2,3], 'value':[10,20,30]})
os.makedirs(raw, exist_ok=True)
df.to_csv(os.path.join(raw,'sample.csv'), index=False)
print('Wrote raw sample to', os.path.join(raw,'sample.csv'))
```

```python
# bronze: basic cleaning
os.makedirs(bronze, exist_ok=True)
df = pd.read_csv(os.path.join(raw,'sample.csv'))
df['value'] = df['value'].astype(float)
df.to_csv(os.path.join(bronze,'sample.csv'), index=False)
print('Wrote bronze CSV')
```

```python
# silver: enrich
os.makedirs(silver, exist_ok=True)
df = pd.read_csv(os.path.join(bronze,'sample.csv'))
df['value_usd'] = df['value'] * 1.0
df.to_csv(os.path.join(silver,'sample_enriched.csv'), index=False)
print('Wrote silver CSV')
```

```python
# gold: aggregate
os.makedirs(gold, exist_ok=True)
df = pd.read_csv(os.path.join(silver,'sample_enriched.csv'))
agg = df['value_usd'].sum()
with open(os.path.join(gold,'summary.txt'),'w') as f:
    f.write(str(agg))
print('Wrote gold summary')
```
